```python
import gradio as gr
```

```python
def reality_checker(statement):
  text=statement.lower()

  high_profile_role=["president of india","prime minister","ceo of google", "ceo of microsoft", "king", "queen"]

  extreme_number=["100", "1000"," 25 hours", " 50 hours", "1 billion"]

  exaggeration_words=["always","never", "everyone","nobody", "best in the world"]


  for role in high_profile_role:
    if role in text:
      return " Unrealistic"

  for num in extreme_number:
    if num in text:
      return "Unrealistic"

  for word in exaggeration_words:
    if word in text:
      return "Exaggerated"

  return "Realistic"



```

```python
gr.Interface(
    fn=reality_checker,
    inputs=gr.Textbox(label="Enter your claim"),
    outputs="text",
    title="AI Claim Reality Checker"
).launch()
```
